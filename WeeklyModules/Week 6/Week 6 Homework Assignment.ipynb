```python
#Read CSV file intoo a dataframe and save as df
df = pd.read_csv("Student Performance Data.csv")
```

```python
# Display first 5 rows in the dataframe
display(df.head)))
```

```python
# Show the original number of rows and columns
print("Original DataFrame:", df.shape)
```

```python
# Create a lambda function that takes a DataFrame and returns
# a new DataFrame with only students whose math scores are 80 or higher
high_math_scores = lambda df: df.loc[df["math score"] >= 80].copy()
```

```python
# Run the lambda function on df and save the new DataFrame
filtered_students = high_math_scores(df)
```

```python
# Display the first five rows of the new DataFrame
display(filtered_students.head())
```

```python
# Show the number of rows and columns in the new DataFrame
print("New DataFrame:", filtered_students.shape)
```
