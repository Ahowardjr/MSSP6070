```python
print(df["EDU_MOTHER"].value_counts(dropna=False))
print(df["EDU_FATHER"].value_counts(dropna=False))
print(df["G_SC"].describe())
```

```python
df.groupby("EDU_MOTHER")["G_SC"].agg(["count", "mean"]).sort_values("mean")
```

```python
df.groupby("EDU_FATHER")["G_SC"].agg(["count", "mean"]).sort_values("mean")
```

```python
mother_means = df.groupby("EDU_MOTHER")["G_SC"].mean()
father_means = df.groupby("EDU_FATHER")["G_SC"].mean()

# Identify the exact keys from the series indices to avoid KeyError due to subtle string differences
mother_primary_key = next((k for k in mother_means.index if k.strip().lower() == 'complete primary'), None)
mother_postgraduate_key = next((k for k in mother_means.index if k.strip().lower() == 'postgraduate education'), None)

father_primary_key = next((k for k in father_means.index if k.strip().lower() == 'complete primary'), None)
father_postgraduate_key = next((k for k in father_means.index if k.strip().lower() == 'postgraduate education'), None)

# Add checks for None in case keys are truly missing
if mother_primary_key is None:
    raise KeyError("Could not find 'Complete primary' in mother's education categories.")
if mother_postgraduate_key is None:
    raise KeyError("Could not find 'Postgraduate education' in mother's education categories.")
if father_primary_key is None:
    raise KeyError("Could not find 'Complete primary' in father's education categories.")
if father_postgraduate_key is None:
    raise KeyError("Could not find 'Postgraduate education' in father's education categories.")

mother_difference = (
    mother_means[mother_postgraduate_key]
    - mother_means[mother_primary_key]
)
father_difference = (
    father_means[father_postgraduate_key]
    - father_means[father_primary_key]
)

print(f"Mother's education difference: {mother_difference:.2f} points")
print(f"Father's education difference: {father_difference:.2f} points")
```

Does parental education influence outcomes, and if so how much?
I compared students’ average scores by their mothers’ and fathers’ education levels. Students whose mothers had postgraduate education scored 20.30 points higher, on average, than students whose mothers had completed primary school. The corresponding difference for fathers’ education was 21.79 points. These results suggest that higher parental education is associated with higher student scores, although the comparison does not establish that parental education caused the difference.

# 2. Which groups/stratum performs best? Are there differences between groups/stratums? Explain your analysis.


```python
df.groupby("STRATUM")["G_SC"].agg(["count", "mean", "std"]).sort_values("mean", ascending = False).round(2)
```

```python
stratum_means = df.groupby("STRATUM")["G_SC"].mean()
stratum_means["Stratum 6"] - stratum_means["Stratum 1"]
```

```python
stratum_means = df[df["STRATUM"].isin(
    [f"Stratum {i}" for i in range(1, 7)]
)].groupby("STRATUM")["G_SC"].mean()

stratum_means = stratum_means.reindex(
    [f"Stratum {i}" for i in range(1, 7)]
)

comparison = pd.DataFrame({
    "average_score": stratum_means.round(2),
    "difference_from_previous": stratum_means.diff().round(2)
})
comparison
```
