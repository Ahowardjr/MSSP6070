```python
# Add a column to the dataframe indicating whether or not the earthquake happened
#in a country or US state that is on the Ring of Fire.
#Use Bolivia, Chile, Ecuador, Peru, Costa Rica, Guatemala, Mexico
#(be sure not to select New Mexico), Japan, Philippines, Indonesia, New Zealand,
# Antarctica (look for the Antarctic), Canada, Fiji, Alaska, Washington,
#California, Russia, Taiwan, Tonga, and the Kermadec Islands).
```

```python
# Store Ring of Fire locations in a list(from arssignment directions)
# Use "Antarctic" to match descriptions containing that word.
ring_locations = [
    "Bolivia", "Chile", "Ecuador", "Peru",
    "Costa Rica", "Guatemala", "Mexico",
    "Japan", "Philippines", "Indonesia",
    "New Zealand", "Antarctic", "Canada",
    "Fiji", "Alaska", "Washington", "California",
    "Russia", "Taiwan", "Tonga", "Kermadec Islands"
]
```

```python
#Create empty list storing True of False for each earthquake(brackets does this)
#True - location matches Ring of Fire list
#False - does not match
ring_results = []
```

```python
#Check each earthquake's location.
#Replace missing locations with empty text so the string checks work.
for place in df["place"].fillna(""):
    on_ring = False

    # Compare the location with each name in our Ring of Fire list.
    for location in ring_locations:
        if location.lower() in place.lower():

            # Skip Mexico as a match if the text says New Mexico.
            if location == "Mexico" and "new mexico" in place.lower():
                continue

            on_ring = True
            break

    # Save this earthquake's True or False result.
    ring_results.append(on_ring)
```

```python
#for place in..checks one row's location at a time
#on_ring = False starts each earthquake as no match
#.lower() makes comparison ignore capitalization
#in checks if a listed name appears in the location text
#continue skips the Mexico match when the text contains New Mexico
#on_ring = True records a match
#break stops checking names once a match is found
#append(on_ring) adds the row's result to ring_results.
```

```python
# Add the True or False results as a new column in the DataFrame.
df["ring_of_fire"] = ring_results

# Display the first five locations alongside their Ring of Fire results.
df[["place", "ring_of_fire"]].head()
```

```python
# Keep rows that are in the Ring of Fire and marked as having a tsunami.
ring_tsunamis = df[
    (df["ring_of_fire"] == True)
    & (df["tsunami"] == 1)]

# Display the number of matching rows.
print(len(ring_tsunamis))
```
