```python
import pandas as pd

# List of extracted stock data files
files = ["aapl.csv", "amzn.csv", "fb.csv", "goog.csv", "nflx.csv"]

dataframes = []

for file in files:
    stock = pd.read_csv(file)
    # Store the ticker in uppercase
    stock["ticker"] = file.replace(".csv", "").upper()
    dataframes.append(stock)

# Combine and display the dataframes
if dataframes:
    combined = pd.concat(dataframes, ignore_index=True)
    display(combined.head())
else:
    print("No dataframes were loaded. Please verify that the CSV files exist in your folder.")
```

```python
# Save the combined dataframe to a CSV file
combined.to_csv('faanf.csv', index=False)

# Verify that the file was created successfully
print("Successfully saved combined data to 'faanf.csv'")
```

 Using 'fanng.csv', use type conversion to change the 'date' column into a 'datetime' and the 'volume' column into integers. Then sort by 'date' and 'ticker.'

```python
import pandas as pd

# Load the dataset from 'faanf.csv'
stocks_df = pd.read_csv('faanf.csv')

# 1. Convert 'date' column to datetime type
stocks_df['date'] = pd.to_datetime(stocks_df['date'])

# 2. Convert 'volume' column to integers
stocks_df['volume'] = stocks_df['volume'].astype(int)

# 3. Sort by 'date' and 'ticker'
stocks_df = stocks_df.sort_values(by=['date', 'ticker']).reset_index(drop=True)

# Display the top rows to verify
display(stocks_df.head(10))
```

Display 7 highest volumes

```python
highest_volume_df = stocks_df.nlargest(7, 'volume')

display(highest_volume_df)
```

Melt the DataFrame:
We use the `melt()` method to make `'date'` and `'ticker'` our identifier variables, combining the remaining columns into a long format.

```python
# Melt the remaining stock metric columns using 'date' and 'ticker' as id_vars
melted_stocks_df = pd.melt(
    stocks_df,
    id_vars=['date', 'ticker'],
    value_vars=['open', 'high', 'low', 'close', 'volume'],
    var_name='metric',
    value_name='value'
)

# Display the melted dataframe
display(melted_stocks_df.head(15))
```
