## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
mean_manual = sum(salaries) / len(salaries)
print(mean_manual)
```

```python
import statistics
mean_verified = statistics.mean(salaries)
print(mean_manual)
```

This means the average salary is $585,690

### median

```python
sorted_salaries = sorted(salaries)
n = len(sorted_salaries)

median_manual = (sorted_salaries[n // 2-1] + sorted_salaries[n // 2]) / 2

print(median_manual)
```

```python

median_verified = statistics.median(salaries)
print(median_verified)
```

This means the middle salary is 589,000. Half of the salaries are above and the other half are below this $589,000.

### mode

```python
mode_manual = max(set(salaries), key=salaries.count)
print(mode_manual)
```

```python
mode_verified = statistics.mode(salaries)
print(mode_verified)
```

The most frequently occuring salary is 477,000

### sample variance
Remember to use Bessel's correction.

Note: Bessel's Correction is n-1

```python
n = len(salaries)

variance_manual = sum((salary-mean_manual) ** 2 for salary in salaries) / (n-1)

print(variance_manual)
```

```python
variance_verified = statistics.variance(salaries)
print(variance_verified)
```

This is a measure of how spread out the salaries are around the mean. Not really sure how beneficial this is here. But this number is expressed in squared dollars and helps us with the next step.

### sample standard deviation
Remember to use Bessel's correction.

```python
standard_deviation_manual = variance_manual ** 0.5
print(standard_deviation_manual)
```

```python
standard_deviation_verified = statistics.stdev(salaries)
print(standard_deviation_verified)
```

This mean the usual distance between the salaries and the mean  of $585,690 is approximately $265,827.

## Exercise 6: Calculating more statistics
### range

```python
minimum_salary = min(salaries)
maximum_salary = max(salaries)
salary_range = maximum_salary - minimum_salary

print("Minimum:", minimum_salary)
print("Maximum:", maximum_salary)
print("Range:", salary_range)
```

Range gives a picture of the spread, but it only uses the most extreme values. Meaning it can be heavily affected by outliers.

### interquartile range

```python
q1 = (sorted_salaries[24] + sorted_salaries[25]) / 2
q3 = (sorted_salaries[74] + sorted_salaries[75]) / 2

interquartile_range = q3 - q1

print("Q1:" , q1)
print("Q3:" , q3)
print("IQR:", interquartile_range)
```

IQR measures the spread of the middle 50% of salaries: IQR = q3 - q1. q1 is center of the lower half, while q3 is the center of the upper half.
Results Meaning:
About 25% of salaries are at or below 402,000
About 25% of salaries are at or above 819,5000
The middle 50% of salaries span approximately $417,500(819,500 - 402,000)

### quartile coefficent of dispersion

```python
quartile_coefficient = (q3 - q1) / (q3 + q1)

print(quartile_coefficient)
print(quartile_coefficient * 100, "%")
```

This means the middle 50% of salaries have a relative dispersion of approximately 34.18%. There's a fair amount of variation between salaries in the middle half of the dataset - the salaries in the middle of the dataset are relatively spread out.

## Exercise 7: Scaling data
### min-max scaling

```python
min_max_scaled = [(salary - minimum_salary) / (maximum_salary - minimum_salary) for salary in salaries]
print(min_max_scaled)
```

```python
print("Minimum scaled value:", min(min_max_scaled))
print("Maximum scaled value:", max(min_max_scaled))
```

Scaling turns each number into a value between 0 and 1; 0 being the lowest and 1 being the highest.

### standardizing

```python
standardized_salaries = [(salary - mean_manual)/ standard_deviation_manual for salary in salaries]
print(standardized_salaries)
```

This tells us how far each salary is from the mean in standard-deviation units.
1 - one standard deviation above the mean; positive values are above the mean
-1 - one standard deviation below the mean; negative values are below the mean

```python
print("Mean:", statistics.mean(standardized_salaries))
print("Standard deviation:", statistics.stdev(standardized_salaries))
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
x_mean = statistics.mean(salaries)
y_mean = statistics.mean(min_max_scaled)
n = len(salaries)

covariance_manual = sum((x-x_mean) * (y-y_mean) for x, y in zip(salaries, min_max_scaled)) / (n-1)

print(covariance_manual)
```

```python
covariance_verified = statistics.covariance(salaries, min_max_scaled)
print(covariance_verified)
```

Covariance indicates whether 2 variables tend to move together:
Positive covariance: they increase together.
Negative covariance: they decrease together.
Near zero: no consistent linear movemnt

### Pearson correlation coefficient ($\rho$)

```python
correlation_manual = covariance_manual / (statistics.stdev(salaries) * statistics.stdev(min_max_scaled))
print(correlation_manual)
```

1.0 correlation means there is a perfect positive relationship.
