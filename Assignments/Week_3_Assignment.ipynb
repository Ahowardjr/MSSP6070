```python
#Define Collatz function
def collatz(number):
  #If number is even divide it by 2 and print results
  if number % 2 == 0:
    print (number//2)
    return number //2
  #If number is odd then multiply by 3 and add 1 and print results
  else:
    print(3 * number + 1)
    return 3 * number + 1

#Ask the user to enter a number

try:
  number = int(input("Enter a number"))
  print(number)
  while number != 1:
    number = collatz(number)
except ValueError:
  print("Must enter a number")
```
